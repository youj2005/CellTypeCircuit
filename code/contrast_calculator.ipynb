```python
import numpy as np
import pickle

cifar_dir = "/storage/ice1/7/8/jyou79/cifar-10-batches-py"
cifar_65_dir = "/storage/ice1/7/8/jyou79/cifar-10-batches-py-65"
```

```python
def rgb_to_grayscale(row):
        R = row[:1024].reshape(32, 32)
        G = row[1024:2048].reshape(32, 32)
        B = row[2048:].reshape(32, 32)
        grayscale = 0.299 * R + 0.587 * G + 0.114 * B
        return grayscale
   
with open(f'%s/data_batch_1'%(cifar_dir), 'rb') as f:
        cifar = pickle.load(f, encoding='bytes') 
cifar = np.array([rgb_to_grayscale(row) for row in cifar[b'data']])
cifar = cifar.astype(np.uint8)
```

```python
stds = np.std(cifar, axis=(1, 2), keepdims=True)
avg_std = np.mean(stds)
std_std = np.std(stds)
means = np.mean(cifar, axis=(1, 2), keepdims=True)
print(avg_std)
print(std_std)
avg_std = 65
```

```python
adjusted_cifar = (cifar - means) / stds * avg_std + means
print(np.std(adjusted_cifar[0]))
```

```python
with open(f'%s/data_batch_1'%(cifar_65_dir), 'wb') as f:
    cifar = pickle.dump(adjusted_cifar, f) 
```
